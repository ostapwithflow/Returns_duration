Просто візуалію данні щоб побачити гістограмму цих дохідностей

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('returns.csv', parse_dates=['open_time'])
df.head()
```

```python
sns.histplot(df['log_return'], bins=200)
plt.show()
```

цікаво,багато нулів,давайте розподіл тривалостей між нулями,нуль це стала.тому має бути кайф 

```python
zeros = df[df['log_return'] == 0]
durations = zeros['open_time'].diff().dt.total_seconds().div(60).dropna()  # хвилини між нулями
durations.describe()
```

```python
sns.histplot(durations, bins=100)
plt.show()
```
